Load dataset


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os

file_path = "/content/sp500_stocks.csv"

df = pd.read_csv(file_path)

print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst five rows:")
display(df.head())

print("\nTotal records:", len(df))
```

Clean dataset and select any one stock


```python
# Standardize column names
df.columns = [
    str(c).strip().lower().replace("_", " ")
    for c in df.columns
]

# Support common dataset column names
date_col = next(
    (c for c in df.columns if c in ["date", "datetime"]),
    None
)

symbol_col = next(
    (c for c in df.columns
     if c in ["symbol", "ticker", "stock"]),
    None
)

close_col = next(
    (c for c in df.columns
     if c in ["close", "closing price", "adj close"]),
    None
)

if date_col is None or symbol_col is None or close_col is None:
    raise ValueError(
        "Required columns not found. Check the column names printed in Step 3."
    )

# Keep only the columns required for this project
data = df[[date_col, symbol_col, close_col]].copy()
data.columns = ["Date", "Symbol", "Close"]

# Clean values
data["Date"] = pd.to_datetime(data["Date"], errors="coerce")
data["Close"] = pd.to_numeric(data["Close"], errors="coerce")
data["Symbol"] = data["Symbol"].astype(str).str.strip().str.upper()

data = data.dropna(subset=["Date", "Close", "Symbol"])
data = data[data["Close"] > 0]
data = data.drop_duplicates(subset=["Date", "Symbol"])
data = data.sort_values(["Symbol", "Date"])

print("Valid total records:", len(data))
print("Number of stocks:", data["Symbol"].nunique())
print("Available date range:",
      data["Date"].min().date(), "to",
      data["Date"].max().date())

# Select Apple
stock = data[data["Symbol"] == "AAPL"].copy()
stock = stock.sort_values("Date").reset_index(drop=True)

if len(stock) < 2000:
    raise ValueError(
        f"AAPL has only {len(stock)} valid records. "
        "Choose another stock with sufficient data."
    )

print("\nSelected stock: AAPL")
print("AAPL records:", len(stock))
print("Start date:", stock["Date"].min().date())
print("End date:", stock["Date"].max().date())

display(stock.head())
```

Plot graph

```python
plt.figure(figsize=(14, 5))
plt.plot(stock["Date"], stock["Close"])
plt.title("Apple (AAPL) Historical Closing Prices")
plt.xlabel("Date")
plt.ylabel("Closing Price (USD)")
plt.grid(True)
plt.tight_layout()
plt.show()
```

Normalise the data


```python
from sklearn.preprocessing import StandardScaler
import numpy as np

# Sort stock data by date
stock = stock.sort_values("Date").reset_index(drop=True)

prices = stock["Close"].values.astype(float)
dates = stock["Date"].values

# Calculate daily percentage returns
returns = np.zeros(len(prices))
returns[1:] = prices[1:] / prices[:-1] - 1

# Split chronologically: 80% training, 20% testing
split = int(len(prices) * 0.80)

# Normalize returns using training data only
scaler = StandardScaler()
scaler.fit(returns[1:split].reshape(-1, 1))

scaled_returns = scaler.transform(
    returns.reshape(-1, 1)
).flatten()

# Generate 60-day sequences
window = 60
X, y, indices = [], [], []

for i in range(window + 1, len(prices)):
    X.append(scaled_returns[i-window:i])
    y.append(scaled_returns[i])
    indices.append(i)

X = np.array(X).reshape(-1, window, 1)
y = np.array(y)
indices = np.array(indices)

# Split sequences chronologically
train_mask = indices < split
test_mask = indices >= split

X_train, y_train = X[train_mask], y[train_mask]
X_test, y_test = X[test_mask], y[test_mask]

print("Training input:", X_train.shape)
print("Testing input:", X_test.shape)
print("Window size:", window)
```

3.Model

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

tf.keras.utils.set_random_seed(42)

model = Sequential([
    Input(shape=(60, 1)),
    LSTM(32),
    Dropout(0.1),
    Dense(16, activation="relu"),
    Dense(1)
])

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.0001),
    loss="mean_squared_error"
)

early_stop = EarlyStopping(
    monitor="val_loss",
    patience=5,
    restore_best_weights=True
)

history = model.fit(
    X_train,
    y_train,
    epochs=30,
    batch_size=32,
    validation_split=0.1,
    shuffle=False,
    callbacks=[early_stop],
    verbose=1
)
```

train history graph


```python
plt.figure(figsize=(9, 4))
plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")
plt.title("LSTM Training and Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Mean Squared Error")
plt.legend()
plt.grid(True)
plt.show()
```

4.Evalute

```python

import numpy as np
import pandas as pd
from sklearn.metrics import mean_absolute_error, mean_squared_error

# Predict next-day returns
pred_scaled = model.predict(X_test, verbose=0)

# Convert scaled returns back to actual returns
pred_returns = scaler.inverse_transform(
    pred_scaled.reshape(-1, 1)
).flatten()

actual_returns = scaler.inverse_transform(
    y_test.reshape(-1, 1)
).flatten()

# Get the original indices of test targets
test_indices = indices[test_mask]

# Convert predicted returns into closing prices
actual = prices[test_indices]
previous_close = prices[test_indices - 1]
predicted = previous_close * (1 + pred_returns)

# Evaluate closing-price predictions
mae = mean_absolute_error(actual, predicted)
rmse = np.sqrt(mean_squared_error(actual, predicted))

print("=" * 45)
print("MODEL EVALUATION RESULTS")
print("=" * 45)
print(f"Mean Absolute Error (MAE): ${mae:.4f}")
print(f"Root Mean Squared Error (RMSE): ${rmse:.4f}")

# Create results using matching date indices
results = pd.DataFrame({
    "Date": dates[test_indices],
    "Actual Close": actual,
    "Predicted Close": predicted
})

display(results.head(10))
```
